```python
#import Libraries

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

#import dataset
df = pd.read_csv("cleaned_cats_dataset.csv")
```

```python
#Line Plot: useful for visualizing data trends over time(time series and numerical data).
plt.plot(df['Age (Years)'], df['Weight (kg)'])
plt.title("Weight vs Age")
plt.xlabel("Age (Years)")
plt.ylabel("Weight (kg)")
plt.show()
```

```python
#Bar plot:used to compare different categories.

df['Breed'].value_counts().plot(kind='bar')
plt.title("Number of Dogs by Breed")
plt.xlabel("Breed")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
#Histogam:used to visualize the distribution of a single numeric variable.

plt.hist(df['Weight (kg)'], bins=10, color='skyblue')
plt.title("Distribution of Weight")
plt.xlabel("Weight (kg)")
plt.ylabel("Frequency")
plt.show()
```

```python
#Box Plot: show the distribution of data based on a five-number summary: minimum, first quartile (Q1), median, third quartile (Q3), and maximum.

sns.boxplot(x="Breed", y="Weight (kg)", data=df)
plt.title("Box Plot of Weight by Breed")
plt.xticks(rotation=45)
plt.show()
```

```python
#scatter plot:show the relationship between two numerical variables.

sns.scatterplot(x="Age (Years)", y="Weight (kg)", hue="Gender", data=df)
plt.title("Scatter Plot of Weight vs Age by Gender")
plt.show()
```

```python
#Pair Plot:great way to visualize relationships between multiple variables in your dataset.

sns.pairplot(df, hue="Gender")
plt.title("Pair Plot of All Features by Gender")
plt.show()
```

```python
#Heatmap:used to visualize the correlation between different variables in the dataset.

correlation = df.corr()
sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title("Heatmap of Feature Correlations")
plt.show()
```
