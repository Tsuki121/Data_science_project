Data Cleaning


## Step 1: Import Necessary Libraries
First, you need to import the libraries that you'll use for data cleaning. Pandas is the most commonly used library for this purpose.


import pandas as pd


# Step 2: Load the CSV File

Next, load your CSV file into a Pandas DataFrame. A DataFrame is like a table where you can easily manipulate and clean data.

[Replace 'your_file.csv' with the actual file path]

df = pd.read_csv('your_file.csv')
This will load the data from the CSV file into the df variable. You can now start cleaning the data.


# Step 3: Preview the Data
Before you start cleaning, it's a good idea to take a look at the first few rows of the data to understand its structure.

print(df.head())

The head() function displays the first five rows of the DataFrame. If you want to see more or fewer rows, you can pass a number to the function, e.g., df.head(10).

# Step 4: Get Basic Information About the Data
You should check for data types, missing values, and overall data structure.


# Display data types and non-null counts
print(df.info())

# Get summary statistics for numerical columns
print(df.describe())
df.info() shows the data types of each column and the number of non-null entries (useful for spotting missing values).
df.describe() gives you summary statistics for numerical columns (mean, standard deviation, min, max, etc.).


# Step 5: Handle Missing Data

5.1 Identify Missing Data
To see how much missing data there is in each column:


print(df.isnull().sum())

5.2 Remove Missing Data
If a column has too many missing values, you might choose to drop it:


df = df.drop(columns=['ColumnName'])
Or, if some rows have missing data in critical columns, you might drop those rows:


df = df.dropna(subset=['ColumnName'])
5.3 Impute Missing Data
If you want to fill missing values, you can use methods like mean, median, or mode:



# Fill missing values with the mean of the column
df['ColumnName'].fillna(df['ColumnName'].mean(), inplace=True)

# Fill missing values with a specific value
df['ColumnName'].fillna(0, inplace=True)
Step 6: Remove Duplicates
If your data contains duplicate rows, you can remove them:

df = df.drop_duplicates()


# Step 7: Handle Outliers

Outliers can distort your analysis. One way to detect outliers is to look at the data distribution using a box plot.

import matplotlib.pyplot as plt

# Box plot to visualize outliers

df.boxplot(column='ColumnName')
plt.show()

To remove outliers, you can use the IQR (Interquartile Range) method:


Q1 = df['ColumnName'].quantile(0.25)
Q3 = df['ColumnName'].quantile(0.75)
IQR = Q3 - Q1

df = df[~((df['ColumnName'] < (Q1 - 1.5 * IQR)) | (df['ColumnName'] > (Q3 + 1.5 * IQR)))]


# Step 8: Convert Data Types

Ensure that all columns have the correct data types. For example, if a column should be numeric but is stored as a string, you can convert it:


df['ColumnName'] = pd.to_numeric(df['ColumnName'], errors='coerce')


# Step 9: Handle Categorical Data

If you have categorical data (e.g., gender, country), convert it to numerical format using encoding techniques:


# One-hot encoding

df = pd.get_dummies(df, columns=['CategoricalColumnName'])

# Step 10: Save the Cleaned Data

After cleaning your data, save it to a new CSV file:


df.to_csv('cleaned_data.csv', index=False)

# Plot

1. Line Plot
Use When: You want to show trends over time or continuous data.
Data Types: Time series data, numerical data.
Example: Plotting weight change over time.

plt.plot(df['Date'], df['Weight'])

plt.title("Weight Over Time")

plt.xlabel("Date")

plt.ylabel("Weight (kg)")

plt.show()


2. Bar Plot
Use When: You want to compare quantities across different categories.
Data Types: Categorical data (x-axis) vs. numerical data (y-axis).
Example: Comparing the number of dogs in different breeds.

df['Breed'].value_counts().plot(kind='bar')

plt.title("Number of Dogs by Breed")

plt.xlabel("Breed")

plt.ylabel("Count")

plt.show()


3. Histogram
Use When: You want to show the distribution of a single numerical variable.
Data Types: Numerical data.
Example: Distribution of dog weights.

plt.hist(df['Weight'], bins=10)

plt.title("Distribution of Weights")

plt.xlabel("Weight (kg)")

plt.ylabel("Frequency")

plt.show()


4. Box Plot
Use When: You want to show the distribution of data based on a five-number summary (min, Q1, median, Q3, max) and detect outliers.
Data Types: Categorical data (x-axis) vs. numerical data (y-axis).
Example: Comparing weight distribution across different breeds.

sns.boxplot(x="Breed", y="Weight", data=df)

plt.title("Box Plot of Weight by Breed")

plt.show()


5. Scatter Plot
Use When: You want to show the relationship between two numerical variables.
Data Types: Numerical data.
Example: Relationship between age and weight.

sns.scatterplot(x="Age", y="Weight", data=df)

plt.title("Scatter Plot of Age vs Weight")

plt.show()


6. Pair Plot
Use When: You want to explore relationships between multiple numerical variables.
Data Types: Multiple numerical data variables.
Example: Relationships between age, weight, and height.

sns.pairplot(df)

plt.title("Pair Plot of All Features")

plt.show()


7. Heatmap
Use When: You want to show the correlation between different variables or display matrix-like data.
Data Types: Numerical data.
Example: Correlation between different features (e.g., age, weight, height).

correlation = df.corr()

sns.heatmap(correlation, annot=True, cmap='coolwarm')

plt.title("Correlation Heatmap")

plt.show()

# How to Decide Which Plot to Use

What is Your Data Type? Determine if your data is numerical, categorical, or time series.

What Do You Want to Show?

Comparisons: Use bar plots or box plots.

Relationships: Use scatter plots or pair plots.

Distribution: Use histograms or box plots.

Trends Over Time: Use line plots.

Correlations: Use heatmaps.

# Examples Based on Specific Needs

To Compare Categories: Bar Plot, Box Plot.

To Explore Relationships: Scatter Plot, Pair Plot.

To Understand Distribution: Histogram, Box Plot.

To Analyze Trends: Line Plot.

To Show Correlations: Heatmap.
