```python
def count(numbers):
    total = 0
    for x in numbers:
        if x < 20:
            total += 1
    return total
```

```python
list_1 = [1,3,7,15,23,43,56,98,17]
count(list_1)
```

```python
nums = [1, 35, 12, 24, 31, 51, 70, 100]

count = 0  # Initialize a counter for values lower than 20
index = 0  # Initialize an index for iterating through the list

while index < len(nums):
    if nums[index] < 20:
        count += 1
    index += 1

print("The number of values lower than 20 is:", count)
```

```python
prices = {
    "box_of_spaghetti" : 4,
    "lasagna" : 5,
    "hamburger" : 2
}

quantity = {
    "box_of_spaghetti" : 6,
    "lasagna" : 10,
    "hamburger" : 0
}

money_spent = 0

for i in prices:
    money_spent = money_spent + (prices[i]*quantity[i])
print(money_spent)
```
